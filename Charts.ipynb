```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
dust_storms = [3000, 2500, 5000, 7000, 7500, 8000, 8500, 8700, 9500, 9800, 9000, 8800]
water_crisis = [2000, 1500, 4000, 6000, 6500, 7000, 7500, 7800, 8500, 8800, 8300, 8100]

plt.figure(figsize=(10,6))
plt.plot(months, dust_storms, marker='o', label='Dust Storms', color='brown')
plt.plot(months, water_crisis, marker='x', label='Water Crisis', color='blue')
plt.xlabel('Month')
plt.ylabel('Number of Tweets')
plt.title('Seasonal Pattern of Environmental Crises')
plt.legend()
plt.grid(True)
plt.show()
```

```python
users = ['User 1', 'User 2', 'User 3', 'User 4', 'User 5']
retweets = [12000, 15000, 18000, 17000, 16000]
mentions = [8000, 10000, 12000, 13000, 11000]

fig, ax = plt.subplots(figsize=(10,6))
width = 0.35  # bar width
x = np.arange(len(users))
ax.bar(x - width/2, retweets, width, label='Retweets', color='skyblue')
ax.bar(x + width/2, mentions, width, label='Mentions', color='orange')

ax.set_xlabel('Users')
ax.set_ylabel('Counts')
ax.set_title('Influence of Key Users: Retweets vs Mentions')
ax.set_xticks(x)
ax.set_xticklabels(users)
ax.legend()
plt.show()
```

```python
emotional_intensity = [0.3, 0.35, 0.4, 0.6, 0.7, 0.75, 0.8, 0.85, 0.9, 1.0, 0.95, 0.95]
sentiment = [68, 22, 10]  # Negative, Neutral, Positive sentiment counts

plt.figure(figsize=(10,6))
plt.scatter(emotional_intensity, sentiment, color='purple')
plt.xlabel('Emotional Intensity (Scale 0-1)')
plt.ylabel('Sentiment Distribution (%)')
plt.title('Correlation Between Emotional Intensity and Sentiment')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
topics = ['Urmia Lake Crisis', 'Air Pollution', 'Zayandeh Rud Water Crisis', 'Dust Storms', 'Climate Change']
tweets = [45000, 40000, 50000, 35000, 21245]

# Plotting the bar chart
plt.figure(figsize=(10,6))
plt.bar(topics, tweets, color=['#3498db', '#e74c3c', '#2ecc71', '#f39c12', '#9b59b6'])

# Title and labels
plt.title('Distribution of Environmental Topics in Tweets', fontsize=16)
plt.xlabel('Environmental Topics', fontsize=14)
plt.ylabel('Number of Tweets', fontsize=14)

# Display the chart
plt.tight_layout()
plt.show()
```

```python
# Pie chart: Distribution of tweets across topics
plt.figure(figsize=(8,8))
plt.pie(tweets, labels=topics, autopct='%1.1f%%', colors=['#3498db', '#e74c3c', '#2ecc71', '#f39c12', '#9b59b6'], startangle=140)
plt.title('Proportion of Tweets Across Environmental Topics', fontsize=16)
plt.axis('equal')  # Equal aspect ratio ensures the pie is drawn as a circle.
plt.show()
```

```python
# Data for stacked bar chart (sample data for three years)
years = ['2021', '2022', '2023']
data = {
    'Urmia Lake Crisis': [12000, 15000, 18000],
    'Air Pollution': [11000, 12000, 17000],
    'Zayandeh Rud Water Crisis': [13000, 14000, 23000],
    'Dust Storms': [8000, 10000, 17000],
    'Climate Change': [7000, 9000, 10450]
}

# Stacked bar chart
plt.figure(figsize=(10,6))
plt.bar(years, data['Urmia Lake Crisis'], label='Urmia Lake Crisis', color='#3498db')
plt.bar(years, data['Air Pollution'], bottom=data['Urmia Lake Crisis'], label='Air Pollution', color='#e74c3c')
plt.bar(years, data['Zayandeh Rud Water Crisis'], bottom=[i+j for i,j in zip(data['Urmia Lake Crisis'], data['Air Pollution'])], label='Zayandeh Rud Water Crisis', color='#2ecc71')
plt.bar(years, data['Dust Storms'], bottom=[i+j+k for i,j,k in zip(data['Urmia Lake Crisis'], data['Air Pollution'], data['Zayandeh Rud Water Crisis'])], label='Dust Storms', color='#f39c12')
plt.bar(years, data['Climate Change'], bottom=[i+j+k+l for i,j,k,l in zip(data['Urmia Lake Crisis'], data['Air Pollution'], data['Zayandeh Rud Water Crisis'], data['Dust Storms'])], label='Climate Change', color='#9b59b6')

# Title and labels
plt.title('Yearly Comparison of Tweet Counts by Environmental Topic', fontsize=16)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Number of Tweets', fontsize=14)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import numpy as np

# Sample correlation data for environmental topics (for illustrative purposes)
corr_data = np.array([
    [1, 0.8, 0.6, 0.5, 0.7],  # Urmia Lake Crisis
    [0.8, 1, 0.75, 0.4, 0.65],  # Air Pollution
    [0.6, 0.75, 1, 0.7, 0.85],  # Zayandeh Rud Water Crisis
    [0.5, 0.4, 0.7, 1, 0.9],    # Dust Storms
    [0.7, 0.65, 0.85, 0.9, 1]   # Climate Change
])

# Labels for topics
corr_topics = topics

# Create the heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_data, xticklabels=corr_topics, yticklabels=corr_topics, cmap='coolwarm', annot=True, fmt='.2f')

# Title and labels
plt.title('Correlation Between Environmental Topics in Tweets', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Sample data over time (months)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
data_monthly = {
    'Urmia Lake Crisis': [1000, 1500, 2000, 2500, 2200, 2700],
    'Air Pollution': [1100, 1400, 1800, 2300, 2100, 2500],
    'Zayandeh Rud Water Crisis': [1200, 1600, 1900, 2400, 2300, 2800],
    'Dust Storms': [800, 1000, 1400, 1700, 1500, 2000],
    'Climate Change': [700, 900, 1200, 1500, 1300, 1600]
}

# Plotting the line graph
plt.figure(figsize=(10,6))
for topic, values in data_monthly.items():
    plt.plot(months, values, label=topic)

# Title and labels
plt.title('Trends of Environmental Topics in Tweets Over Time', fontsize=16)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Number of Tweets', fontsize=14)
plt.legend()

plt.tight_layout()
plt.show()
```
