```python
import pandas as pd
import re

# Assuming you have a DataFrame called df with a column 'Tweet'
# Create a new column 'hashtags' to store the extracted hashtags
df['hashtags'] = None

# Define a function to extract hashtags from a given tweet
def extract_hashtags(tweet):
    hashtags = re.findall(r'#(\w+)', tweet)
    return hashtags

# Apply the function to each row in the 'Tweet' column and store the result in the 'hashtags' column
df['hashtags'] = df['Tweet'].apply(extract_hashtags)

# Display the updated DataFrame
df.head()
```

**Hashtag Analysis**

```python
# Flatten the list of hashtags
all_hashtags = [tag for sublist in df['hashtags'] for tag in sublist]

# Count the frequency of each hashtag
hashtag_counts = pd.Series(all_hashtags).value_counts()

# Display the top hashtags
print("Top Hashtags:")
print(hashtag_counts.head())
```

```python
import pandas as pd
import re
from collections import Counter

# Define a function to extract hashtags from a text
def extract_hashtags(text):
    hashtags = re.findall(r'#\w+', text)
    return hashtags

# Apply the function to your dataset to extract hashtags from all tweets
df['hashtags'] = df['Tweet'].apply(extract_hashtags)

# Flatten the list of hashtags into a single list
all_hashtags = [tag for sublist in df['hashtags'] for tag in sublist]

# Count the occurrences of each hashtag using Counter
hashtags_counts = Counter(all_hashtags)

# Convert the Counter object to a DataFrame for easier export
hashtags_df = pd.DataFrame(hashtags_counts.items(), columns=['Hashtag', 'Count'])

# Sort the DataFrame by hashtag count (descending order)
hashtags_df = hashtags_df.sort_values(by='Count', ascending=False)

# Export the DataFrame to a CSV file
hashtags_df.to_csv('hashtags_counts.csv', sep=',', index=False, encoding='utf-8-sig')
```

**User Engagement**

```python
# User Metrics

user_metrics = df.groupby('ID').agg({'Retweet': 'sum', 'Likes': 'sum', 'Tweet': 'count'}).reset_index()

# Influence Score (simple sum of retweets and likes)
user_metrics['influence_score'] = user_metrics['Retweet'] + user_metrics['Likes']

# Display user metrics
print("\nUser Engagement Metrics:")
print(user_metrics.sort_values(by='influence_score', ascending=False))
```

**Retweet and Like Analysis**

```python
# Identify the most retweeted and liked tweets
top_retweets = df.sort_values(by='Retweet', ascending=False).head()
top_likes = df.sort_values(by='Likes', ascending=False).head()

# Display the top retweeted and liked tweets
print("\nTop Retweeted Tweets:")
print(top_retweets[['ID', 'Tweet', 'Retweet']])

print("\nTop Liked Tweets:")
print(top_likes[['ID', 'Tweet', 'Likes']])
```

**Mention Analysis**

```python
# Identify frequently mentioned users
mentions = [mention for sublist in df['Tweet'].str.findall(r'@(\w+)') for mention in sublist]
mention_counts = pd.Series(mentions).value_counts()

# Display frequently mentioned users
print("\nFrequently Mentioned Users:")
print(mention_counts.head())
```

**Time-based Engagement**

```python
# Assuming you have a timestamp column in your DataFrame
df['Date'] = pd.to_datetime(df['Date'])

# Time-based engagement analysis using the 'Date' column
time_based_engagement = df.set_index('Date').resample('D').agg({'Retweet': 'sum', 'Likes': 'sum'})

# Display time-based engagement
print("\nTime-based Engagement:")
print(time_based_engagement)
```

```python
import re
import networkx as nx
import matplotlib.pyplot as plt

# Assuming your DataFrame has a column 'text' containing the tweet text
# (Replace this with your actual data)
#df

# Function to extract mentions from a text
def extract_mentions(text):
    return re.findall(r'@(\w+)', text)

# Apply the function to create a new column 'mentions'
df['mentions'] = df['Tweet'].apply(extract_mentions)

# Create a DataFrame for edges (source_user, target_user)
edges_list = []

# Iterate over rows to create edges DataFrame
for _, row in df.iterrows():
    for mention in row['mentions']:
        edges_list.append({'source_user': row['ID'], 'target_user': mention})

# Create a DataFrame from the edges list using concat
edges = pd.concat([pd.DataFrame(edge, index=[0]) for edge in edges_list], ignore_index=True)

# Create a directed graph using NetworkX
G = nx.from_pandas_edgelist(edges, 'source_user', 'target_user', create_using=nx.DiGraph())

# Visualization
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, font_size=8, node_size=800, node_color='skyblue', edge_color='gray', font_color='black', font_weight='bold', arrowsize=10)
plt.title('User Interaction Network')
plt.show()
```

```python
# Visualize a subset of nodes (top 100 by degree centrality)
top_nodes = sorted(nx.degree_centrality(G).items(), key=lambda x: x[1], reverse=True)[:100]
subset_G = G.subgraph([node[0] for node in top_nodes])

# Visualization with adjusted layout and color
plt.figure(figsize=(12, 10))
pos = nx.spring_layout(subset_G)
nx.draw(subset_G, pos, with_labels=False, font_size=8, node_size=800, node_color='skyblue', edge_color='gray', font_color='black', font_weight='bold', arrowsize=10)
plt.title('Top 100 Users by Degree Centrality in Interaction Network')
plt.show()
```

```python
import pandas as pd
import networkx as nx
import community.community_louvain as cl
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Assuming 'data' is your DataFrame with columns 'ID' and 'Tweet'

# Create a directed graph
G = nx.Graph()

# Extract unique user IDs and tweets from the dataset
unique_users = df['ID'].unique()
unique_tweets = df['Tweet'].unique()

# Create nodes for users and tweets
G.add_nodes_from(unique_users, node_type='user')
G.add_nodes_from(unique_tweets, node_type='tweet')

# Create edges based on Jaccard Similarity
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(unique_tweets)
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)


for i, user_id in enumerate(unique_users):
    for j, tweet_text in enumerate(unique_tweets):
        similarity = cosine_sim[i, j]
        if similarity > 0.9:  # Adjust the similarity threshold as needed################################################
            G.add_edge(user_id, tweet_text, weight=similarity)

# Apply the Louvain method for community detection
partition = cl.best_partition(G)

# Add community information as a node attribute
nx.set_node_attributes(G, partition, 'community')

# Visualize the network with communities
plt.figure(figsize=(12, 10))
pos = nx.spring_layout(G)
cmap = cm.get_cmap('viridis', max(partition.values()) + 1)
node_size = [len(G.neighbors(node)) * 50 for node in G.nodes()]  # Adjust the multiplier for node size
nx.draw(G, pos, with_labels=True, font_size=8, node_size=node_size, node_color=[partition[node] for node in G.nodes()], cmap=cmap, edge_color='gray', font_color='black', font_weight='bold', arrowsize=10)
plt.title('Network with Community Detection (Jaccard Similarity)')
plt.show()
```

```python
#RUN  on Kaggle############## It needs more RAM
import pandas as pd
import networkx as nx
import community.community_louvain as cl
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# Create a directed graph
G = nx.Graph()

# Extract unique user IDs and tweets from the dataset
unique_users = df['ID'].unique()
unique_tweets = df['Tweet'].unique()

# Create nodes for users and tweets
G.add_nodes_from(unique_users, node_type='user')
G.add_nodes_from(unique_tweets, node_type='tweet')

# Create edges based on Jaccard Similarity
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(unique_tweets)
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Set the number of top similarities to visualize
top_n = 10

for i, user_id in enumerate(unique_users):
    similarities = [(j, cosine_sim[i, j]) for j in range(len(unique_tweets))]
    similarities.sort(key=lambda x: x[1], reverse=True)

    for j, similarity in similarities[:top_n]:
        G.add_edge(user_id, unique_tweets[j], weight=similarity)

# Apply the Louvain method for community detection
partition = cl.best_partition(G)

# Add community information as a node attribute
nx.set_node_attributes(G, partition, 'community')

# Visualize the network with communities
plt.figure(figsize=(12, 10))

# Use the Kamada-Kawai layout for better node placement
pos = nx.kamada_kawai_layout(G)

# Extract user nodes and their communities for labeling
user_nodes = [node for node in G.nodes if G.nodes[node]['node_type'] == 'user']
user_communities = [partition[node] for node in user_nodes]

# Draw user nodes with user names as labels
nx.draw_networkx_nodes(G, pos, nodelist=user_nodes, node_size=800, node_color=user_communities, cmap=cm.get_cmap('viridis', max(partition.values()) + 1))
nx.draw_networkx_labels(G, pos, labels={node: node for node in user_nodes}, font_size=8, font_color='black')

# Draw edges
nx.draw_networkx_edges(G, pos, width=0.5, alpha=0.5, edge_color='gray')

plt.title('Network with Community Detection (Top-N Similarities)')
plt.show()

```
